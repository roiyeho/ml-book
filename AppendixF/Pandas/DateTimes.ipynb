```python
import pandas as pd
```

```python
df = pd.read_csv('data/flights_sample_3m.csv')
```

Converting to DateTime

```python
df['FL_DATE'] = pd.to_datetime(df['FL_DATE'])
```

```python
df['FL_DATE'] = pd.to_datetime(df['FL_DATE'], format='%Y-%m-%d')
```

DateTime Comparisons

```python
print(df[(df['FL_DATE'] >= '2022-01-01') & (df['FL_DATE'] <= '2022-01-03')])
```

Datetime Attributes

```python
df['FL_YEAR'] = df['FL_DATE'].dt.year          
df['FL_MONTH'] = df['FL_DATE'].dt.month       
df['FL_DAY'] = df['FL_DATE'].dt.day
df['FL_WEEKDAY'] = df['FL_DATE'].dt.day_name()

print(df[['FL_DATE', 'FL_MONTH', 'FL_DAY', 'FL_WEEKDAY']])
```

```python
df.groupby(df['FL_DATE'].dt.floor('D')).size()
```

```python
df = df.set_index('FL_DATE')
print(df.loc['2022-06-01'])  # all flights from June 1, 2022
```

```python
print(df.loc['2022-06'])  # all flights from June 2022
```

```python
df = df.sort_index()
print(df.loc['2022-06-01':'2022-06-05']) 
```
